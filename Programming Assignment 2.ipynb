# A. REPRODUCIBLE NORMALIZATION PROBLEM

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
print(X)
```

```python
Mean = np.mean(X)
print(Mean)
```

```python
Standard_Deviation = np.std(X)
print(Standard_Deviation)
```

```python
X_normalized = (X - Mean) / Standard_Deviation
print(X_normalized)


```

```python
print(np.mean(X_normalized))
```

```python
print(np.std(X_normalized))
```

```python
np.save("X_normalized.npy", X_normalized)
```

# B. CUBES DIVISIBLE BY 4 PROBLEM

```python
C = np.arange(1,101,1)
C
```

```python
C = C**3
C
```

```python
C = np.reshape(C,(10,10))
C
```

```python
print(np.shape(C))
```

```python
print(C)
```

```python
div_by_4 = C[C % 4 == 0]
div_by_4
```

```python
print(div_by_4)
```

```python
print(len(div_by_4))
```

```python
np.save("div_by_4.npy", div_by_4)
```

# C. ABOVE-MEAN SQUARES PROBLEM

```python
S = np.arange(1,37,1)
S
```

```python
S = S**2
S
```

```python
S = np.reshape(S,(6,6))
S
```

```python
print(S)
```

```python
S_mean = np.mean(S)
S_mean
```

```python
print(S_mean)
```

```python
above_mean = S[S_mean < S]
above_mean
```

```python
print(above_mean)
```

```python
print(len(above_mean))
```

```python
np.save("above_mean.npy", above_mean)
```
